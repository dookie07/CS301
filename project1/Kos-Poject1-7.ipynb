```python
#read the file
df = pd.read_csv("./ProjectDataSet.csv")
```

```python
#rename the columns for clarity and easier access
df.rename(columns={
    'geo_type': 'GeoType',
    'geo_area_id': 'GeoAreaID',
    'geo_area_name': 'GeoAreaName',
    'borough_id': 'BoroughID',
    'time_period': 'Year',
    'Children tested for lead by age 3 years Number': 'TestedNumber',
    'Children tested for lead by age 3 years Number _NOTES': 'TestedNumberNotes',
    'Children tested for lead by age 3 years Percentage': 'TestedPercentage',
    'Children tested for lead by age 3 years Percentage _NOTES': 'TestedPercentageNotes'
}, inplace=True)
```

```python
df.head()
```

```python
#check what type is each column
df.dtypes
```

```python
# Drop notes columns (metadata not needed for analysis)
df.drop(columns=['TestedNumberNotes', 'TestedPercentageNotes'], inplace=True)
```

```python
# Convert numerical columns to correct types
df['TestedNumber'] = pd.to_numeric(df['TestedNumber'], errors='coerce')
df['TestedPercentage'] = pd.to_numeric(df['TestedPercentage'], errors='coerce')
```

```python
#Display missing value summary
missing_summary = df.isnull().sum()
print("Missing Values per Column:\n", missing_summary)
```

```python
# Fill missing values in numerical columns with the mean
df['TestedNumber'].fillna(df['TestedNumber'].mean(), inplace=True)
df['TestedPercentage'].fillna(df['TestedPercentage'].mean(), inplace=True)
```

```python
# Fill missing values in categorical columns with mode (most frequent)
df['BoroughID'].fillna(df['BoroughID'].mode()[0], inplace=True)
```
