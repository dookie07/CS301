```python
import pandas as pd
import numpy as np
import scipy.stats as stats
from scipy.stats import norm
```

```python
mean= 40
sd = 15
n =100
minValue = 38
maxValue = 44
sqrt_mean = np.sqrt(n)
```

```python
#getting z scores values
zMin =  (minValue - mean) / (sd/sqrt_mean)
zMax = (maxValue - mean) / (sd/sqrt_mean)
print("Min z score: " , zMin)
print("Max z core: " , zMax)
```

```python
#Probability of the z scores
minProb = norm.cdf(zMin)
maxProb = norm.cdf(zMax)
finProb = maxProb - minProb
print("The final  probability that the sample mean is between 38 and 44 is of" ,finProb)
```
